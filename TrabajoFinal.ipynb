```python
import pandas as pd

ruta = r"C:\Users\andry\Downloads\base-saij-normativa-provincial.csv"
df = pd.read_csv(ruta)

df.head()
```

Filtrar los datos solo para las provincias de Tucumán, Formosa y Tierra del Fuego.


```python
provincias = ["Tucumán", "Formosa", "Tierra del Fuego"]

df_filtrado = df[df["provincia_nombre"].isin(provincias)]

df_filtrado
```

Limpiar columnas vacías o que no aporten información valiosa para su uso.

```python
# Crear una copia para conservar el dataframe original
df_limpio = df_filtrado.copy()

# Eliminar columnas completamente vacías
df_limpio = df_limpio.dropna(axis=1, how="all")

# Eliminar columnas redundantes
df_limpio = df_limpio.drop(columns=["provincia_id"], errors="ignore")

# Mostrar las columnas resultantes
df_limpio.info()

df_limpio.head()
```

Filtrar datos de normativas de Ley, Decreto Ley y Texto Ordenado Ley.

```python
tipos_normativa = ["Ley", "Decreto Ley", "Texto Ordenado Ley"]

df_normativas = df_limpio[
    df_limpio["tipo_norma"].isin(tipos_normativa)
].copy()

df_normativas
```

Eliminar casos de normativas vetadas.


```python
# Eliminar normativas cuyo estado indique que fueron vetadas
df_sin_vetadas = df_normativas[
    ~df_normativas["estado_vigencia"]
    .fillna("")
    .str.contains("vetad", case=False, na=False)
].copy()

df_sin_vetadas
```

Reemplazar vacíos por “n/d”

```python
df_final = df_sin_vetadas.copy()

# Reemplazar valores nulos y cadenas vacías por "n/d"
df_final = df_final.fillna("n/d")
df_final = df_final.replace(r"^\s*$", "n/d", regex=True)

df_final
```

Guardar en .xlsx y .csv (UTF-8 separador “,”).

```python
!pip install openpyxl
```

```python
# Guardar el dataframe final en formato XLSX y CSV
df_final.to_excel("normativas_filtradas.xlsx", index=False)
df_final.to_csv(
    "normativas_filtradas.csv",
    index=False,
    sep=",",
    encoding="utf-8"
)

print("Archivos guardados correctamente.")
```
